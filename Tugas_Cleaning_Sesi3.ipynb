# Tugas Data Cleaning Sesi 3 - Fnhabibi

```
import csv
seen=set()
rows=[]
with open('data_kotor.csv') as f:
    import csv
    r=csv.DictReader(f)
    for row in r:
        if row['Id'] in seen: continue
        seen.add(row['Id'])
        rows.append(row)
print(f'Data kotor: {len(rows)} setelah hapus duplikat')
```
